#### 🎵 Day 10 Mini Project: Playlist Manager Using Doubly Linked Lists
This project shows how doubly linked lists help manage data that needs bidirectional navigation, like music playlists.

📀 Playlist Features
Each song has:

Title
Artist

✅ Supported Operations:

Add a song to the playlist
Delete a song by title
Search and update song info
Display songs forward and backward
Reverse the playlist

```python
class SongNode:
    def __init__(self, title, artist):
        self.title = title
        self.artist = artist
        self.next = None
        self.prev = None

class PlaylistManager:
    def __init__(self):
        self.head = self.tail = None

    def display_forward(self):
        print(f'\n🎶 Playlist (Forward):', end="\n")
        current = self.head

        while current:
            print(f'- {current.title} by {current.artist}')
            current = current.next

    
    def add_song(self, title, artist):
        new_node = SongNode(title, artist)

        if self.head is None:
            self.head = self.tail = new_node

        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node
        print(f'✅ Added song: {title} by {artist}')


    def display_backward(self):
        print(f'\n🎶 Playlist (Backward):', end="\n")
        if self.tail is None:
            print(f'❌ Playlist is empty!')

        current = self.tail
        while current:
            print(f'- {current.title} by {current.artist}')
            current = current.prev


    def search_and_update(self, title, new_title=None, new_artist=None):
        if self.head is None:
            print(f'❌ Playlist is empty!')
            return

        current = self.head
        while current:
            if current.title == title:
                print(f'\n🔍 Found song: {current.title} by {current.artist}')
                if new_title:
                    print(f'✏️ Updated song: {current.title} → {new_title}')
                    current.title = new_title
                if new_artist:
                    print(f'✏️ Updated song: {current.artist} → {new_artist}')
                    current.artist = new_artist
                return
            current = current.next
          
        print(f'❌ Song not in playlist!!')

    
    def delete_song(self, title):
        current = self.head
        if current is None:
            print(f'❌ Playlist is empty, Nothing to delete!')
            return False
          
        while current:
            if current.title == title:
                print(f'\n🗑️ Deleted song: {title}')

                # Case 1: Node is 
                if current == self.head:
                    self.head = current.next
                    if self.head:
                        self.head.prev = None
                    else:
                        self.tail = None   # set self.tail = None to reflect that the list is now empty (if its the only node in the list)

                # Case 2: Node is tail
                elif current == self.tail:
                    self.tail = current.prev
                    self.tail.next  = None

                # Case 3: Node is in the middle
                else:
                    current.prev.next = current.next
                    current.next.prev = current.prev

                return True

            current = current.next

        print(f'❌ Song not in playlist!!')
        return False
    

    def reverse_playlist(self):
        current = self.head

        while current:
            current.next, current.prev = current.prev, current.next

            current = current.prev
        
        self.head, self.tail = self.tail, self.head
        print(f'\n🔁 Playlist Reversed')
        self.display_backward()
      




a = PlaylistManager()
a.add_song("Shape of You", "Ed Sheeran")
a.add_song("Blinding Light", "The Weekend")
a.add_song("Levitating", "Dua Lipa")
a.search_and_update("Blinding Light", "Save Your Tears")
a.display_forward()
a.display_backward()
a.reverse_playlist()
a.delete_song("Shape of You")

```
